```python
customers_df = spark.createDataFrame(customers_data, ["customer_id","customer_name","city","age"])
```

```python
products_df = spark.createDataFrame(products_data, ["product_id","product_name","category","price","stock"])
```

```python
orders_df = spark.createDataFrame(orders_data, ["order_id","customer_id","product_id","quantity","total_amount","order_date"])
```

```python
orders_df = orders_df.withColumn("order_date", col("order_date").cast("date"))
```

```python
sales_df = orders_df.join(customers_df, "customer_id").join(products_df, "product_id")
```

```python
final_sales_report = sales_df.select("customer_name","city","product_name","category",col("price").alias("product_price"),"quantity","total_amount","order_date")
```
